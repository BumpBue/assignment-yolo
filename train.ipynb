```python
import os
import shutil
import yaml
import torch
from ultralytics import YOLO

print(f"PyTorch version: {torch.__version__}")
print("พร้อมใช้งาน Ultralytics YOLOv8")

device = "cpu"
print(f"Device set to: {device}")
```

```python
data_yaml_path = 'dataset/data.yaml' 

if os.path.exists(data_yaml_path):
    with open(data_yaml_path, 'r', encoding='utf8') as f:
        yaml_info = yaml.safe_load(f)
        
    print(f"เจอไฟล์ {data_yaml_path} แล้ว!")
    print(f"จำนวนคลาส (nc): {yaml_info.get('nc')}")
    print(f"รายชื่อคลาส (names): {yaml_info.get('names')}")
else:
    print(f"หาไฟล์ไม่พบ กรุณาเช็คที่อยู่ไฟล์: {data_yaml_path}")
```

```python
print("--- กำลังเริ่มเทรนโมเดล Polygon (Segmentation) ---")

model_polygon = YOLO('yolov8n-seg.pt')

results_polygon = model_polygon.train(
    data=data_yaml_path,
    epochs=60,
    patience=15,
    batch=8,
    imgsz=416,
    device='cpu',
    workers=4,
    name='train_polygon',
    
    mosaic=1.0,
    degrees=10.0,
    translate=0.1,
    scale=0.5, 
    fliplr=0.5,
    flipud=0.0,
    hsv_h=0.015, hsv_s=0.7, hsv_v=0.4
)

print("เทรนโมเดล Polygon เสร็จสิ้น!")
print("ไฟล์โมเดลถูกบันทึกไว้ที่: runs/segment/train_polygon/weights/best.pt")
```

```python
print("--- กำลังเริ่มเทรนโมเดล Rectangle (Object Detection) ---")

model_rectangle = YOLO('yolov8n.pt')

results_rectangle = model_rectangle.train(
    data=data_yaml_path,
    epochs=60,
    patience=15,
    batch=8,
    imgsz=416,
    device='cpu',
    workers=4,
    name='train_rectangle',
    
    mosaic=1.0,
    degrees=10.0,
    translate=0.1,
    scale=0.5,
    fliplr=0.5,
    flipud=0.0,
    hsv_h=0.015, hsv_s=0.7, hsv_v=0.4
)

print("เทรนโมเดล Rectangle เสร็จสิ้น!")
print("ไฟล์โมเดลถูกบันทึกไว้ที่: runs/detect/train_rectangle/weights/best.pt")
```

```python
src_polygon = 'runs/segment/train_polygon/weights/best.pt'
dst_polygon = 'best_polygon.pt'
if os.path.exists(src_polygon):
    shutil.copy(src_polygon, dst_polygon)
    print(f"คัดลอก {dst_polygon} สำเร็จ")

src_rectangle = 'runs/detect/train_rectangle/weights/best.pt'
dst_rectangle = 'best_bbox.pt'
if os.path.exists(src_rectangle):
    shutil.copy(src_rectangle, dst_rectangle)
    print(f"คัดลอก {dst_rectangle} สำเร็จ")

print("\nเตรียมไฟล์โมเดลเสร็จสมบูรณ์! คุณสามารถนำ best_polygon.pt และ best_bbox.pt ไปใส่ใน GitHub และรันบนเว็บได้เลย")
```
